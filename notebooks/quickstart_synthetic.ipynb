# ContinualRAG-BEIR quickstart (synthetic data)

Demonstrates the API end to end **without downloads or GPUs**.

> **Plumbing demo only.** The synthetic pools are clustered Gaussians, not real BEIR text, so the numbers here are *not*
> expected to match the paper (e.g. K-Center may look fine). Use `scripts/run_benchmark.py` for the real benchmark.

````
import pandas as pd
from continual_rag import Config, run_suite
from continual_rag.data import make_synthetic_pools
from continual_rag.policies import EXT8_SET
from continual_rag.report import summary_table, print_verdicts
from continual_rag.audit import run_audit
````

````
cfg = Config(domains=list("abcd"), docs_per_domain=300, max_queries_per_domain=60,
             n_usage_per_phase=40, cap=480)           # cap = 40% of 1,200 streamed docs
pools = make_synthetic_pools(n_domains=4, n_docs=400, n_queries=80, seed=1)
````

## Run policies over 10 seeds (paired episodes)

````
df = run_suite(cfg, pools, range(10), sorted(EXT8_SET), verbose=False)
summary_table(df).round(4)
````

## Pre-declared hypothesis tests (H_a, H_b, H_c, H_d, H_k)

````
print_verdicts(df, 'synthetic demo')
````

## Pre-eviction predictability audit

````
run_audit(cfg, pools, range(5)).round(3)
````
